```python
import random
import matplotlib.pyplot as plt

# ==========================================
# 1. DATA INPUT & PARAMETER ALGORITMA GENETIKA
# ==========================================

courses = [
    ("AI", "DosenA"), ("ML", "DosenB"), ("DB", "DosenC"),
    ("CN", "DosenD"), ("SE", "DosenE"), ("OS", "DosenF"),
    ("HCI", "DosenE"), ("IR", "DosenA"), ("DL", "DosenB"),
    ("NLP", "DosenB"), ("CV", "DosenC"), ("IOT", "DosenD"),
    ("CyberSec", "DosenF"), ("BigData", "DosenA"), ("NOSQL", "DosenC"),
    ("Cloud", "DosenD"), ("UX", "DosenE"), ("SecEng", "DosenF"),
    ("DevOps", "DosenA"), ("RL", "DosenB"), ("ETL", "DosenC"),
    ("Agile", "DosenD"), ("GameDev", "DosenE"), ("Blockchain", "DosenF")
]

rooms = ["R1", "R2", "R3", "R4"]

timeslots = [
    "Senin-08", "Senin-10", "Senin-13",
    "Selasa-08", "Selasa-10", "Selasa-13",
    "Rabu-08", "Rabu-10", "Rabu-13",
    "Kamis-08", "Kamis-10", "Kamis-13"
]

POPULATION_SIZE = 60
GENERATIONS = 100
CROSSOVER_RATE = 0.85
MUTATION_RATE = 0.20

# ==========================================
# 2. FUNGSI EVALUASI KONFLIK & FITNESS
# ==========================================

def calculate_conflicts(individual):
    """
    Menghitung total konflik berdasarkan 3 aturan:
    1. Konflik Ruang: Ruang & timeslot sama untuk MK berbeda.
    2. Konflik Dosen-Waktu: Dosen mengajar >1 MK di timeslot sama.
    3. Konflik Dosen-Hari: Dosen mengajar >1x di hari yang sama.
    """
    conflicts = 0

    room_time_set = set()
    lecturer_time_set = set()
    lecturer_day_set = set()

    for gen in individual:
        course, lecturer, room, time = gen
        day = time.split("-")[0]

        # 1. Konflik Ruang
        room_time_key = (room, time)
        if room_time_key in room_time_set:
            conflicts += 1
        else:
            room_time_set.add(room_time_key)

        # 2. Konflik Dosen-Waktu
        lecturer_time_key = (lecturer, time)
        if lecturer_time_key in lecturer_time_set:
            conflicts += 1
        else:
            lecturer_time_set.add(lecturer_time_key)

        # 3. Konflik Dosen-Hari
        lecturer_day_key = (lecturer, day)
        if lecturer_day_key in lecturer_day_set:
            conflicts += 1
        else:
            lecturer_day_set.add(lecturer_day_key)

    return conflicts

def fitness_score(individual):
    """
    Menghitung nilai fitness skala 0 - 100.
    Rumus: 100 / (1 + total_konflik)
    """
    conflicts = calculate_conflicts(individual)
    return 100.0 / (1.0 + conflicts)

# ==========================================
# 3. INISIALISASI & OPERATOR GENETIKA
# ==========================================

def create_individual():
    individual = []
    for course, lecturer in courses:
        room = random.choice(rooms)
        time = random.choice(timeslots)
        individual.append((course, lecturer, room, time))
    return individual

def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, len(courses) - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
        return child1, child2
    return parent1[:], parent2[:]

def mutate(individual):
    mutated_ind = []
    for gen in individual:
        course, lecturer, room, time = gen
        if random.random() < MUTATION_RATE:
            # Ubah ruang atau timeslot secara acak
            if random.random() < 0.5:
                room = random.choice(rooms)
            else:
                time = random.choice(timeslots)
        mutated_ind.append((course, lecturer, room, time))
    return mutated_ind

def selection(population):
    # Tournament Selection
    tournament = random.sample(population, 3)
    tournament.sort(key=lambda ind: fitness_score(ind), reverse=True)
    return tournament[0]

# ==========================================
# 4. PROSES EVOLUSI (MAIN LOOP)
# ==========================================

population = [create_individual() for _ in range(POPULATION_SIZE)]

max_fitness_history = []
avg_fitness_history = []

for gen in range(1, GENERATIONS + 1):
    fitnesses = [fitness_score(ind) for ind in population]

    max_fit = max(fitnesses)
    avg_fit = sum(fitnesses) / POPULATION_SIZE

    max_fitness_history.append(max_fit)
    avg_fitness_history.append(avg_fit)

    # Cetak perkembangan tiap generasi
    print(f"Gen {gen:2d}: Max Fitness = {max_fit:6.2f} | Avg Fitness = {avg_fit:6.2f}")

    # Buat generasi baru
    new_population = []

    # Elitisme: Pertahankan individu terbaik
    best_ind = population[fitnesses.index(max_fit)]
    new_population.append(best_ind)

    while len(new_population) < POPULATION_SIZE:
        parent1 = selection(population)
        parent2 = selection(population)

        child1, child2 = crossover(parent1, parent2)

        new_population.append(mutate(child1))
        if len(new_population) < POPULATION_SIZE:
            new_population.append(mutate(child2))

    population = new_population

# ==========================================
# 5. MENAMPILKAN HASIL TERBAIK & EVALUASI
# ==========================================

# Ambil individu terbaik dari generasi terakhir
fitnesses = [fitness_score(ind) for ind in population]
best_schedule = population[fitnesses.index(max(fitnesses))]
best_fitness = max(fitnesses)
remaining_conflicts = calculate_conflicts(best_schedule)

print("\n=== JADWAL TERBAIK ===")
for gen in best_schedule:
    course, lecturer, room, time = gen
    print(f"MK: {course:<9} | Ruang: {room} | Waktu: {time:<8} | Dosen: {lecturer}")

print("\n=== HASIL EVALUASI AKHIR ===")
print(f"Fitness Terbaik: {best_fitness:.2f}")
print(f"Jumlah Konflik : {remaining_conflicts}")

# ==========================================
# 6. VISUALISASI GRAFIK EVOLUSI
# ==========================================

plt.figure(figsize=(10, 5))
plt.plot(range(1, GENERATIONS + 1), max_fitness_history, label="Fitness Maksimum", color="#1f77b4", linewidth=2)
plt.plot(range(1, GENERATIONS + 1), avg_fitness_history, label="Fitness Rata-rata", color="#ff7f0e", linewidth=2)

plt.title("Perkembangan Fitness - Algoritma Genetika Penjadwalan Mata Kuliah", fontsize=12)
plt.xlabel("Generasi", fontsize=10)
plt.ylabel("Fitness", fontsize=10)
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
